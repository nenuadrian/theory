```python
import numpy as np
import matplotlib.pyplot as plt
print('NumPy', np.__version__, '— ready.')
```

```python
# A 2-16-1 MLP trained from scratch on two interleaving moons.
rng = np.random.default_rng(0)
m = 120
t = np.linspace(0, np.pi, m)
X = np.vstack([np.c_[np.cos(t), np.sin(t)],
               np.c_[1 - np.cos(t), 0.5 - np.sin(t)]]) + rng.normal(0, 0.10, (2*m, 2))
Y = np.r_[np.zeros(m), np.ones(m)].reshape(-1, 1)

sigmoid = lambda z: 1 / (1 + np.exp(-z))
W1 = rng.normal(0, 0.8, (2, 16)); b1 = np.zeros((1, 16))
W2 = rng.normal(0, 0.8, (16, 1)); b2 = np.zeros((1, 1))

lr, losses = 0.6, []
for epoch in range(2500):
    a1 = np.tanh(X @ W1 + b1)
    p = sigmoid(a1 @ W2 + b2)
    losses.append(-np.mean(Y*np.log(p+1e-7) + (1-Y)*np.log(1-p+1e-7)))
    dz2 = (p - Y) / len(X)
    dW2 = a1.T @ dz2; db2 = dz2.sum(0, keepdims=True)
    dz1 = (dz2 @ W2.T) * (1 - a1**2)
    dW1 = X.T @ dz1; db1 = dz1.sum(0, keepdims=True)
    W1 -= lr*dW1; b1 -= lr*db1; W2 -= lr*dW2; b2 -= lr*db2

print('final loss', round(losses[-1], 4), '— accuracy', f'{((p>0.5)==Y).mean():.0%}')
plt.plot(losses); plt.title('training loss'); plt.xlabel('epoch'); plt.grid(alpha=.3); plt.show()
```
